## Abstract

Our project studies whether NYPD officers of different racial groups receive civilian complaints at different rates compared to their representation in the police force. As a group we wanted to better understand whether officers of certain races receive more or fewer complaints relative to how many officers of that race work in the NYPD.  We also chose this topic because it is a very important issue and we had the data to analyze it.

To explore this question, we used multiple public datasets including the Civilian Complaint Review Board Complaints dataset and the NYPD Personnel Demographics dataset. We also analyzed the NYPD arrest data set and New York City population demographic data set to provide additional context about different policing patterns in the city.

First we cleaned and organized the datasets, followed by conducting exploratory data analysis to identify trends in both complaints and officer demographics. Next we compared the percentage of complaints received by officers across different races to the percentage of officers in the NYPD representing those same racial groups. We then analyzed arrest data. We compared arrest rates by race with the racial distribution of New York City's population to gain a deeper understanding of broader policing patterns that might affect complaint trends.

Our findings and results indicate that the distribution of complaints does not correlate perfectly with the racial distribution of the NYPD. It was found that White officers received the highest overall number of complaints and made up a larger share of complaints than their proportion within the police force. In contrast, officers from other racial groups often see complaint proportions that are approximately equal to or slightly less than their representation in the department. Our analysis of arrest data also revealed that Black and Hispanic individuals are arrested at rates exceeding their population share in New York City, whereas White and Asian individuals experience arrest rates that fall below their population percentages.

These results point to complex relationships between officer demographics, policing actions, and community interactions regarding both civilian complaints and policing outcomes. It's important to note that the complaint data only represents reported cases and not substantiated misconduct, so these results shouldn't be seen as direct reflections of officer conduct. Instead, this analysis points to patterns that may be indicative of larger social and structural dynamics affecting policing and how police and the community interact.

### Dataset #1 - CCRB

The CCRB (Civilian Complaint Review Board) dataset contains records of civilian complaints filed against NYPD officers. Each row represents one closed complaint case and includes information about the officer involved, the type of allegation, the date of the incident, and the final disposition of the investigation. Allegation type describes the nature of the complaint, such as abuse of authority, use of force, or offensive language. Incident date is recorded as a calendar date and indicates when the complaint occurred. Disposition describes the outcome of the CCRB investigation and typically falls into categories such as substantiated, unsubstantiated, or exonerated. These metrics allow us to quantify complaint patterns across officer racial groups.

There are several important limitations to this dataset. Complaint data reflects reported incidents, not confirmed misconduct. Some misconduct may go unreported due to many reasons, some of which include lack of trust in reporting systems, or lack of awareness of the complaint process. Reporting rates may also vary across neighborhoods and communities, which introduces potential reporting bias. Additionally, some fields contain missing values, particularly in demographic related columns. Because of these concerns, complaint counts should not be interpreted as direct measures of officer behavior but rather as reported interactions that triggered formal complaints.

### Dataset #4 - NYC Population Demographics (ACS)
The NYC Population Demographics dataset contains demographic breakdowns for New York City's population at both the citywide and borough levels. This dataset contains approximately 8.5 million residents with demographic estimates across 6 regions and approximately 90 categories for each region.

The important variables are the race proportions: White, Black, Asian, Hispanic, American Indian, and Pacific Islander. These are provided as both counts and percentages for each borough and for NYC as a whole. The race categories are categorical variables that represent the racial composition of NYC's population.

The purpose of this dataset is to establish expected proportions for arrest analysis.The problems with this dataset include that all proportions and estimates are not exact and are given with margins of error, as they come from survey estimates rather than exact counts. Additionally, the dataset represents a snapshot from the 2012-2016 ACS 5-year estimates, which may not reflect current population demographics if used for more recent analyses.

```python
raw = pd.read_csv('data/00-raw/NYC_Population_Demographics.csv', header=None)
# Construct column names for each borough and metric combination
regions = ['Bronx', 'Brooklyn', 'Manhattan', 'Queens', 'Staten Island', 'New York City']
metrics = ['Estimate', 'Margin of Error', 'Percent', 'Percent MoE']
col_names = ['Subject', 'Subcategory']
for r in regions:
    for m in metrics:
        col_names.append(f'{r}_{m}')
df_acs = raw.iloc[5:].copy()
df_acs.columns = col_names[:df_acs.shape[1]]
df_acs['Subject'] = df_acs['Subject'].ffill()

# Clean dataset by remove metadata rows, assign column names, and fill missing subject labels
nyc_cols = [c for c in df_acs.columns if c.startswith('New York City_')]
def to_numeric_clean(s):
    v = str(s).strip().replace(',', '').replace('+/-', '').replace('%', '').replace('*****', '').replace('(X)', '')
    try:
        return pd.to_numeric(v) if v and v != 'nan' else np.nan
    except Exception:
        return np.nan
for c in nyc_cols:
    df_acs[c] = df_acs[c].apply(to_numeric_clean)
df_acs = df_acs[df_acs[nyc_cols].notna().any(axis=1)]

# Convert NYC columns to numeric by stripping symbols and invalids
df_nyc = df_acs[['Subject', 'Subcategory'] + nyc_cols].copy()
df_nyc = df_nyc.rename(columns={c: c.replace('New York City_', 'nyc_').replace(' ', '_') for c in nyc_cols})


print("Shape (tidy NYC-only):", df_nyc.shape)
df_nyc.head()
```

```python
# Missingness check
missing = df_nyc.isna().sum()
print("Missing counts per column:\n", missing)
print("\nRows with any missing:", df_nyc.isna().any(axis=1).sum())
```

```python
# Outliers
print("nyc_Estimate range:", df_nyc['nyc_Estimate'].min(), "–", df_nyc['nyc_Estimate'].max())
print("nyc_Percent range:", df_nyc['nyc_Percent'].min(), "–", df_nyc['nyc_Percent'].max())

outlier = df_nyc[(df_nyc['nyc_Percent'].notna()) & ((df_nyc['nyc_Percent'] < 0) | (df_nyc['nyc_Percent'] > 100))]
print("Rows with percent outside [0,100]:", len(outlier))
outlier_est = df_nyc[(df_nyc['nyc_Estimate'].notna()) & (df_nyc['nyc_Estimate'] < 0)]
print("Rows with negative estimate:", len(outlier_est))
```

```python
# Clean
numeric_cols = ['nyc_Estimate', 'nyc_Margin_of_Error', 'nyc_Percent', 'nyc_Percent_MoE']
nyc_df = df_nyc[df_nyc[numeric_cols].notna().any(axis=1)].copy()

print("Cleaned shape:", nyc_df.shape)
nyc_df.head()
```

#### Section 2 of EDA: Arrest Disparities Relative to NYC Population

In this section, we look into the relationships between NYC population demographics, NYPD arrest data, and NYPD officer demographics. We want to assess whether racial disparities exist in NYPD arrest patterns relative to the racial composition of NYC's population, and examine how the the NYPD's demographic compares to the communities they serve.

To investigate these disparities, we compare the racial distribution of arrests by the NYPD with the racial composition of NYC's population. Using the data from the NYPD Historic Arrests dataset and population estimates from the American Community Survey, we calculate each racial group's total arrests and compare it to the city's population. We then can get a disparity ratio that indicates whether certain groups are over or under represented in arrests relative to their population size

A disparity ratio greater than 1.0 indicates that a group is arrested at a higher rate than their population share would expect, while a ratio below 1.0 means the opposite. This analysis helps our research question because it sets a baseline understanding of how race interacts with policing outcomes in NYC, which helps us identify patterns in civilian complaints.

We begin by examining the cleaned NYC population demographics and the arrest counts by race.

```python
nyc_df
```

```python
arrest_large_df.groupby(by = 'PERP_RACE').count()
```

```python
arrest_proportions = (arrest_large_df["PERP_RACE"].value_counts(normalize=True).reset_index())
arrest_proportions.columns = ["Race", "Arrest Proportion"]

arrest_proportions.set_index("Race")["Arrest Proportion"].plot(kind="bar")
plt.title("Proportion of Arrests by Race")
plt.ylabel("Proportion of Arrests")
plt.show()

print(arrest_proportions)
```

The bar chart above shows the proportion of total NYPD arrests attributed to each racial group in the historic arrests dataset. **Black** individuals account for the largest share of arrests at approximately 48.7%, followed by **Hispanic** individuals at 34.3%. **White** individuals represent 12.3% of arrests, while **Asian** and **American Indian** individuals account for 4.4% and 0.2%, respectively. To interpret whether these proportions represent over or under representation, we need to compare them against the actual racial composition of New York City's population, shown below.

```python
nyc_proportions = nyc_df[nyc_df["Subject"] != "Total population"][["Subject", "nyc_Percent"]]
nyc_proportions.columns = ["Race", "Population Proportion"]
nyc_proportions["Population Proportion"] = (nyc_proportions["Population Proportion"]/100)

nyc_proportions.set_index("Race")["Population Proportion"].plot(kind="bar")
plt.title("NYC Population Proportion by Race")
plt.ylabel("Proportion of Population")
plt.show()

print(nyc_proportions)
```

When comparing the two bar charts above we find that **White** residents form the largest demographic group in NYC at 45.3%, they make up only 12.3% of arrests. On the other hand, **Black** residents represent 26% of the population but account for a disproportionate 48.7% of all arrests. **Hispanic** residents show a smaller but still notable gap being 29.0% of the population and 34.3% of arrests. To quantify these differences precisely, we merge the two proportion datasets and compute a **disparity ratio**.

```python
difference_df = arrest_proportions.merge(nyc_proportions, on="Race")
difference_df["Disparity Ratio"] = (difference_df["Arrest Proportion"]/difference_df["Population Proportion"])
difference_df
```

```python
difference_df.set_index("Race")["Disparity Ratio"].plot(kind="bar")
plt.axhline(1.0, linestyle="--")
plt.title("Racial Arrest Disparity")
plt.ylabel("Arrests VS Population")
plt.show()
```

The dashed line at 1.0 represents where a group being arrested at exactly the rate their population share would be.

- **Black** individuals have a disparity ratio of approximately **1.87**, meaning they are arrested at almost twice the rate that their share of the NYC population.
- **Hispanic** individuals have a disparity ratio of approximately **1.18**, indicating a moderately high arrest rate relative to their population share.
- **White** individuals have a disparity ratio of approximately **0.27**, meaning they are arrested at a far lower rate than their population proportion would predict. Despite having a 45.3% share of the city's population.
- **Asian** and **American Indian** groups similarly show disparity ratios well below 1.0 at 0.29 and 0.23 indicating substantially lower arrest rates relative to their population shares.

### Chi-Square Analysis #1: Racial Disparities in Arrest Rates

Our EDA reveals the difference between how racial groups are represented in the NYC population and how often they appear in arrest records. To determine whether these differences are **statistically significant**, we apply a **chi-square goodness-of-fit test**.

This test compares the **observed distribution** of arrests across racial groups to the **expected distribution** derived from each group's share of the NYC population. If arrests were proportional to population, the observed and expected counts would be similar. A large chi-square statistic and a correspondingly small p-value would indicate that the differences are unlikely due to chance alone. We standardized both distributions to a common scale of 100,000 to make the comparison meaningful.

**Null Hypothesis (H₀)**: Arrest proportions match population proportions for each racial group.  
**Alternative Hypothesis (Hₐ)**: Arrest proportions differ significantly from population proportions.

```python
# Creating the Observed VS Expected Counts
observed_prop = difference_df["Arrest Proportion"].values
expected_prop = difference_df["Population Proportion"].values

expected_prop = expected_prop/expected_prop.sum()

# Standarized At 100,000 people
observed = observed_prop * 100000
expected = expected_prop * 100000

chi_stat, p_val = chisquare(f_obs=observed, f_exp=expected)

print("Chi_stat:", chi_stat)
print("Df:", len(observed) - 1)
print("p-val:", p_val)
```

The chi-square goodness of fit test reveals a test statistic of **χ² ≈ 58,830** with **4 degrees of freedom** and a p-value of **0.0**. This gives us overwhelming evidence to **reject the null hypothesis** that arrest proportions mirror population proportions across racial groups.

The racial distribution of arrests in NYC is inconsistent with what we would expect it to be if every racial group were arrested at a rate proportional to its population share. To find out which groups drive this disparity, we analyze the **standardized residuals**, which measure the direction and magnitude of each group's deviation from expectation in standard deviation.

```python
# Computing the Standarized Residuals
std_res=(observed-expected)/np.sqrt(expected)
difference_df["Std Residual"] = std_res

print(difference_df[["Race", "Std Residual"]])
```

```python
# Plotting the Residuals
plt.bar(difference_df["Race"], difference_df["Std Residual"])

plt.axhline(y=0, linestyle="--")

plt.title("Standardized Residuals by Race")
plt.ylabel("Residuals")
plt.xticks(rotation=45)
plt.show()
```

The standardized residuals quantify how many standard deviations each group's observed arrest count falls from its expected count. Values above and below **±2** are statistically significant.

- **Black** individuals have the largest positive residual (**≈ 176.2**). This means that they are arrested at a higher rate than their population share would predict.
- **Hispanic** individuals have a positive residual (**≈ 59.4**), meaning that they are overly arrested relative to their population.
- **White** individuals have a large negative residual (**≈ −135.0**), meaning that they are arrested at a much lower rate than their population proportion.
- **Asian** individuals have a strong negative residual (**≈ −74.7**), revealing that they are also arreseted at a lower rate.
- **American Indian** individuals have a moderately negative residual (**≈ −21.5**), however it is important to note that this group is a very small share of both the population and arrests.

These residuals confirm that the significant chi-square result is not driven by a single outlier group.  Next, we decompose the overall chi-square statistic into each group's individual contribution to see which groups carry the most statistical weight.

```python
# Finding and Plotting the Chi-Square Contributions
chi_contributions = ((observed-expected)**2)/expected

difference_df["Chi-Square Contribution"] = chi_contributions

bars = plt.bar(
    difference_df["Race"],
    difference_df["Chi-Square Contribution"]
)

for i in bars:
    height = i.get_height()
    plt.text(i.get_x() + i.get_width() / 2, height, f"{height:.2f}",ha='center', va='bottom')

plt.title("Chi-Square Contribution by Race")
plt.xlabel("Race")
plt.ylabel("Contribution to Chi Stat")

plt.show()
```

The chi-square contribution chart shows the overall test statistic down by racial group. The **Black** group contributes the most to the chi-square value, followed by **White** and **Asian** groups. This reveals that the most significant individual from arrest rates occurs for Black individuals. They are arrested at far higher rates than their population share would predict, while White and Asian individuals are arrested at rates below their population shares.

These results provide strong statistical evidence that **arrest rates in NYC are not racially proportional to the population**. Black and Hispanic New Yorkers face high arrest rates, while White and Asian residents are clearly underrepresented in arrest data relative to their demographics.

### Chi-Square Analysis #2: Officer Race and Officer Complaints

In our second analysis we will find out if an officer's race has any relation to the outcome of civilian complaints filed against them.

If the complaint process is shared equally, then we would expect complaint outcomes to be **independent** of officer ethnicity. So, the distribution of complaints should look roughly the same regardless of whether the officer's race.

We test this using a **chi-square test of independence** on a contingency table that cross tabulates officer ethnicity against board disposition. This test evaluates whether two categorical variables are statistically associated with each other.

**Null Hypothesis (H₀)**: Officer ethnicity and complaint outcomes are independent  
**Alternative Hypothesis (Hₐ)**: They are dependent

```python
# Creating the Chi-Square test of independence
df = ccrb_filtered.dropna(subset=["mos_ethnicity", "board_disposition"])

observed = pd.crosstab(df["mos_ethnicity"], df["board_disposition"])

chi2, p_val, df, expected = chi2_contingency(observed)

print("Chi_stat:", chi2)
print("Df:", df)
print("p-val:", p_val)
```

The chi-square test of independence produces a test statistic of **χ² ≈ 143.57** with **40 degrees of freedom** and a p-value of **1.36 × 10⁻¹³**. This small p-value means that we **reject the null hypothesis** as there is a statistically significant association between officer ethnicity and the outcome of civilian complaints.

This means that the disposition a complaint receives is **not independent** of the officer's race. Officers of different race appear to experience systematically different complaint outcomes. We now examine the **standardized residuals** in a heatmap, which reveals which ethnicity combinations deviate most from what independence would predict. 

```python
# Creating the Standarized Residuals from the chi-square test
expected = pd.DataFrame(expected, index=observed.index, columns=observed.columns)

std_res =(observed-expected)/np.sqrt(expected)

plt.figure(figsize=(12,6))

sns.heatmap(std_res, annot=True, cmap="coolwarm", center=0)
plt.title("Standardized Residuals for Officer Ethnicity vs Complaint Outcome")
plt.show()
```

The standardized residuals heatmap reveals the specific ethnicity disposition pairings that deviate most from the pattern expected under independence. Cells shaded in dark red indicate combinations that happen **more often** than expected, while cells shaded in blue happen **less often** than expected.

The heatmap shows that:

- **White officers** tend to have a higher than expected rate of complaints that are "Substantiated," meaning that complaints against White officers are more likely to be upheld by the CCRB relative to what we would expect if race didn't play a role.
- **Black and Hispanic officers** show different residual patterns across disposition categories, revealing that the complaint process does not have identical outcome distributions for all officer groups.
- Both strong positive and negative residuals across multiple groups confirms that the chi-square statistic is not driven by a single cell but reflects a broad pattern of association between officer race and complaint outcome.

```python
# Creating the Chi-test contributions 
chi_contributions =((observed-expected)**2)/expected

plt.figure(figsize=(12,6))

sns.heatmap(chi_contributions, annot=True, cmap="coolwarm")

plt.title("Chi-Stat Contributions for Officer Ethnicity vs Complaint Outcome")
plt.show()
```

The chi-square contribution heatmap above shows the cells that have the greatest influence on the overall test statistic. The largest contributions reveal the specific officer race and disposition combinations. These high contribution cells have the most impact on the statistically significant result.


Across both tests, **race is a statistically significant factor** in both civilian and officer parts of NYPD policing outcomes. Analysis 1 demonstrates how arrest rates are not proportional to population demographics, as Black and Hispanic New Yorkers are arrested at disproportionately higher rates. Analysis 2 reveals that officer ethnicity and complaint disposition are not independent, with complaint outcomes varying systematically by officer race.

### Random Forest Regression Analysis 3: Predicting Complaint Rates from Officer Demographics

The chi-square tests above reveal that officer race is associated with complaint outcomes, but they do not tell us how well demographic features can predict complaint rates, nor which features matter most. So we also train a **Random Forest Regression** model that uses officer demographics, race, gender, rank, and group size to predict the **complaint rate** for each demographic.

We use a pipeline with one-hot encoding for the categorical features (race, gender, rank) and pass `num_officers` through as a numeric feature.

```python
# Cleaning dataframes by standardizing gender and rank
def simplify_rank(rank):
    if rank in ["Police Officer", "Detective", "Sergeant", "Lieutenant", "Captain", "Inspector", "Deputy Inspector"]:
        return rank
    else:
        return "Other"
ccrb_filtered["rank_now"] = ccrb_filtered["rank_now"].apply(simplify_rank)
officer_filtered["ranklit"] = officer_filtered["ranklit"].apply(simplify_rank)

ccrb_filtered["mos_gender"] = ccrb_filtered["mos_gender"].replace({"M": "Male","F": "Female"})
officer_filtered = officer_filtered[officer_filtered["gender"].isin(["Male", "Female"])]  
```

```python
# Grouping up each df by race, gender, and rank by complaints
complaints = (ccrb_filtered.groupby(["mos_ethnicity", "mos_gender", "rank_now"]).size().reset_index(name="num_complaints"))
officers = (officer_filtered.groupby(["race", "gender", "ranklit"]).size().reset_index(name="num_officers"))

# Standarizing column names in complaints
complaints = complaints.rename(columns={"mos_ethnicity": "race","mos_gender": "gender","rank_now": "ranklit"})
```

```python
# Creating a derived metric, complaint rate, from merging the dfs
df = officers.merge(complaints,on=["race", "gender", "ranklit"],how="left")

df["num_complaints"] = df["num_complaints"].fillna(0)
df["complaint_rate"] = df["num_complaints"] / df["num_officers"]
```

The cells above prepare the modeling dataset. We first standardize officer ranks into a manageable set of categories, then aggregate both the CCRB complaint records and the NYPD personnel data by the same grouping keys of race, gender, and rank. After merging these two aggregations, each row represents a unique demographic subgroup with its officer count and complaint count. The variable, **complaint_rate**, is computed as the number of complaints divided by the number of officers, giving us a normalized measure of complaint burden per officer.

Below, we define the feature matrix and preprocessing pipeline, then train and evaluate the Random Forest model.

```python
# Prepping the Regression Model
X = df[["race", "gender", "ranklit", "num_officers"]]
y = df["complaint_rate"]

cats = ["race", "gender", "ranklit"]

preprocess = ColumnTransformer(transformers=[
        ("cat", OneHotEncoder(handle_unknown="ignore"), cats)],
        remainder="passthrough"
        )
```

```python
# Adding parameters, fitting model and creating predictions
forest = RandomForestRegressor(n_estimators=200,max_depth=10)
pipe = Pipeline([("preprocess", preprocess),("rf", forest)])

X_train, X_test, y_train, y_test = train_test_split(X, y, test_size=0.2)
pipe.fit(X_train, y_train)

y_pred = pipe.predict(X_test)
```

```python
print("MSE:", mean_squared_error(y_test, y_pred))
print("R2:", r2_score(y_test, y_pred))
```

The **Mean Squared Error ≈ 1.45** indicates that, on average, the model's predicted complaint rates deviate from actual rates by roughly the square root of 1.45 (≈ 1.20). The **R² ≈ −0.69** is negative, meaning the model performs *worse* than simply predicting the mean complaint rate for every subgroup.

A negative R² does not mean the model is useless for our analysis, but it tells us that **officer demographics alone are not enough to accurately predict complaint rates**. Complaint rates likely depend on additional factors not captured in our data, such as assignment location, shift patterns, years of experience, type of duty, or neighborhood-level crime and policing intensity. However, even when a model has poor predictive power, its **feature importance scores** can still reveal which of the included variables the model relied on most heavily when making splits, which gives us insight into the influence of race versus other demographic features.

```python
# Finding out which features are the most important specifically in relation to race
feats = pipe.named_steps["preprocess"].get_feature_names_out()
imp = pipe.named_steps["rf"].feature_importances_

imp_df = pd.DataFrame({"Feature": feats,"Importance": imp}).sort_values(by="Importance", ascending=False)
race_imp_df = imp_df[imp_df['Feature'].str.contains('race')]
```

We extract the feature importance scores from the trained Random Forest and filter for **race related features** specifically, since our research question centers on whether officer race influences complaint patterns. The bar chart below shows which race categories the model weighted most heavily when constructing its decision trees.

```python
sns.barplot(data=race_imp_df.head(5), x="Importance", y="Feature")

plt.title("Top Features By Race Influencing Complaint Rate")
plt.show()
```

The feature influence plot shows which race related one hot encoded features the Random Forest relied on most when splitting nodes. Although the model's overall predictive accuracy is low, the relative ranking of feature importances still provides directional insight. The race categories that appear with higher importance are the ones where complaint rates vary most across subgroups which is consistent with the chi-square findings above.
